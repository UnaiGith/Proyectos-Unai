#2018

```python
import fastf1
import pandas as pd

# Define the season
seasons = [2018]  # Changed to a list containing only 2018

# Create an empty list to store lap data
all_laps = []

# Loop through each season (in this case, only 2018)
for season in seasons:
    # Get the schedule for the season
    schedule = fastf1.get_event_schedule(season)

    # Loop through each event
    for _, event in schedule.iterrows():
        try:
            # Load the race session for each event
            session = fastf1.get_session(season, event['EventName'], 'R')  # 'R' for Race session
            session.load()

            # Get lap data and add additional info
            laps = session.laps
            laps['EventName'] = event['EventName']
            laps['Season'] = season
            all_laps.append(laps)

            print(f"Processed: {season} - {event['EventName']}")
        except Exception as e:
            print(f"Could not process {season} - {event['EventName']}: {e}")

# Combine all lap data into a single DataFrame
laps_df = pd.concat(all_laps, ignore_index=True)

# Save to CSV
laps_df.to_csv('/content/laps2018.csv', index=False)
print("DataFrame created and saved!")
```

```python
import pandas as pd

# Cargar datos
df = pd.read_csv("./Datos_definitivos.csv")

# Ordenar para facilitar la detección de pitstops
df = df.sort_values(by=["Driver", "Season", "EventName", "LapNumber"])
```

###Detectamos las Pitstops

```python
# Detectar pitstops si cambia el compuesto o reinicia la vida del neumático
df["PitstopDetected"] = (
    (df["TyreLife"] == 1) |
    (df["Compound"] != df.groupby(["Driver", "Season", "EventName"])["Compound"].shift(1))
)
```

###Filtro los datos

```python
# Convertimos 'Rainfall' a booleano explícito
df["Rainfall"] = df["Rainfall"].astype(str).str.lower().isin(["true", "1", "yes"])

# Eliminar vueltas con lluvia y las primeras 2 vueltas
df = df[(df["Rainfall"] == False) & (df["LapNumber"] > 2)]

# Filtrar pilotos que terminaron en el top 10
max_lap = df.groupby(["Driver", "Season", "EventName"])["LapNumber"].transform("max")
last_laps = df[df["LapNumber"] == max_lap]
top10 = last_laps.groupby(["Driver", "Season", "EventName"])["Position"].last().reset_index()
top10 = top10[top10["Position"] <= 10]

# Quedarnos solo con los pilotos del top 10
df = df.merge(top10[["Driver", "Season", "EventName"]], on=["Driver", "Season", "EventName"], how="inner")
```

###Tiempo a segundos

```python
df["LapTime"] = pd.to_timedelta(df["LapTime"])
df["Sector1Time"] = pd.to_timedelta(df["Sector1Time"])
df["Sector2Time"] = pd.to_timedelta(df["Sector2Time"])
df["Sector3Time"] = pd.to_timedelta(df["Sector3Time"])

df["LapTimeSec"] = df["LapTime"].dt.total_seconds()
df["Sector1Sec"] = df["Sector1Time"].dt.total_seconds()
df["Sector2Sec"] = df["Sector2Time"].dt.total_seconds()
df["Sector3Sec"] = df["Sector3Time"].dt.total_seconds()
```

###Codificación de categorías

```python
from sklearn.preprocessing import LabelEncoder
import joblib
categorical_cols = ["Driver", "Team", "Compound"]
label_encoders = {col: LabelEncoder().fit(df[col]) for col in categorical_cols}

joblib.dump(label_encoders["Driver"], "enc_driver.pkl")
joblib.dump(label_encoders["Team"], "enc_team.pkl")
joblib.dump(label_encoders["Compound"], "enc_compound.pkl")


for col in categorical_cols:
    df[col + "_enc"] = label_encoders[col].transform(df[col])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar el archivo CSV
df = pd.read_csv("Datos_definitivos.csv")

# Asegúrate de que las columnas necesarias existan y tengan los tipos de datos correctos
# (especialmente 'LapNumber', 'Compound', y la lógica para detectar el fin de un stint)

# --- Adaptación para detectar el fin de un stint ---
# Asumiendo que quieres considerar un cambio de compuesto como el fin de a stint.
# Si tienes otra lógica (como una columna 'PitstopDetected'), úsala en su lugar.
# Aquí creamos una columna booleana que es True si el compuesto cambia respecto a la vuelta anterior
# Usamos .transform() instead of .apply() to ensure the index aligns with the original DataFrame
df['Compound_Changed'] = df.groupby(['Driver', 'Season', 'EventName'])['Compound'].transform(lambda x: x.ne(x.shift()))

# El inicio de un nuevo stint es cuando Compound_Changed es True
# También el inicio del primer stint para cada grupo (Driver, Season, EventName)
# We need to ensure the cumsum is applied within each group as well for stint counting
df['StintStart'] = df.groupby(['Driver', 'Season', 'EventName'])['Compound_Changed'].cumsum()


# Si tu archivo 'Datos_definitivos.csv' ya tiene una columna que indica pitstops o stints,
# puedes usar esa columna directamente en lugar de la lógica anterior.
# Por ejemplo, si tienes una columna 'Pitstop' donde 1 indica un pitstop:
# df['StintStart'] = df.groupby(['Driver', 'Season', 'EventName'])['Pitstop'].cumsum()


# Agrupar por StintStart y Compound para calcular la duración de cada stint
# Agrupamos por Driver, Season, EventName para asegurarnos de que los stints son específicos de cada piloto y evento
stint_durations = df.groupby(['Driver', 'Season', 'EventName', 'StintStart', 'Compound'])['LapNumber'].agg(['min', 'max']).reset_index()
stint_durations['StintLength'] = stint_durations['max'] - stint_durations['min'] + 1 # Duración en vueltas

# Calcular la duración media del stint por compuesto
avg_stint_length_by_compound = stint_durations.groupby('Compound')['StintLength'].mean().sort_values(ascending=False)

# Calcular la frecuencia de uso de cada compuesto (contando cuántas veces aparece en el DataFrame original)
compound_usage = df['Compound'].value_counts()

# Visualizar la duración media del stint por compuesto
plt.figure(figsize=(10, 6))
sns.barplot(x=avg_stint_length_by_compound.index, y=avg_stint_length_by_compound.values)
plt.title('Duración Media del Stint por Compuesto')
plt.xlabel('Compuesto')
plt.ylabel('Duración Media del Stint (Vueltas)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Visualizar la frecuencia de uso de cada compuesto
plt.figure(figsize=(10, 6))
sns.barplot(x=compound_usage.index, y=compound_usage.values)
plt.title('Frecuencia de Uso de Compuestos')
plt.xlabel('Compuesto')
plt.ylabel('Número de Vueltas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# O un gráfico de pastel para la frecuencia de uso
plt.figure(figsize=(8, 8))
plt.pie(compound_usage, labels=compound_usage.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribución del Uso de Compuestos')
plt.axis('equal') # Asegura que el pastel sea circular.
plt.show()
```

###Función Base EDNN

```python
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Embedding, Flatten, Concatenate, Dropout
from tensorflow.keras.optimizers import RMSprop, Nadam
from tensorflow.keras.losses import MeanSquaredError
loss = MeanSquaredError()

def build_ednn_model(
    num_inputs,
    embedding_inputs,
    embedding_dims,
    hidden_units=[150, 100],
    dropout_rate=0.3,
    output_type='regression'
):
    # Inputs categóricos (embeddings)
    embedding_layers = []
    inputs = []

    for name, (input_dim, output_dim) in embedding_inputs.items():
        inp = Input(shape=(1,), name=f"{name}_input")
        emb = Embedding(input_dim=input_dim, output_dim=output_dim, name=f"{name}_emb")(inp)
        emb = Flatten()(emb)
        embedding_layers.append(emb)
        inputs.append(inp)

    # Inputs numéricos
    num_input = Input(shape=(num_inputs,), name="numeric_input")
    inputs.append(num_input)

    # Concatenar todo
    x = Concatenate()(embedding_layers + [num_input])

    # Capas ocultas
    for units in hidden_units:
        x = Dense(units, activation="relu")(x)
        x = Dropout(dropout_rate)(x)

    # Capa de salida
    if output_type == 'regression':
        output = Dense(1, activation="linear")(x)
        loss = tf.keras.losses.MeanSquaredError()
        metrics = ["mae","accuracy"]
        optimizer = RMSprop(learning_rate=0.001)
    else:  # clasificación
        output = Dense(1, activation="sigmoid")(x)
        loss = "binary_crossentropy"
        metrics = ["accuracy"]
        optimizer = Nadam(learning_rate=0.001)

    model = Model(inputs=inputs, outputs=output)
    model.compile(optimizer=optimizer, loss=loss, metrics=metrics)
    return model
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    precision_score,
    recall_score,
    f1_score,
)
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestRegressor

# 1. Predicted vs Actual (Regresión)
def plot_predicted_vs_actual_positions(y_true, y_pred):
    plt.figure(figsize=(8, 6))
    plt.scatter(y_true, y_pred, alpha=0.4, edgecolor="k")
    plt.plot([min(y_true), max(y_true)], [min(y_true), max(y_true)], "r--", label="Ideal")
    plt.xlabel("Actual Position")
    plt.ylabel("Predicted Position")
    plt.title("Predicted vs Actual Driver Position")
    plt.legend()
    plt.grid(True)
    plt.show()

# 2. Matriz de Confusión
def plot_confusion_matrix(y_true, y_pred_binary):
    cm = confusion_matrix(y_true, y_pred_binary)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["No Pitstop", "Pitstop"])
    disp.plot(cmap="Blues")
    plt.title("Confusion Matrix - Pitstop Prediction")
    plt.grid(False)
    plt.show()

# 3. Métricas de Clasificación
def plot_classification_metrics(y_true, y_pred_binary):
    precision = precision_score(y_true, y_pred_binary)
    recall = recall_score(y_true, y_pred_binary)
    f1 = f1_score(y_true, y_pred_binary)

    metrics = {"Precision": precision, "Recall": recall, "F1 Score": f1}
    plt.figure(figsize=(6, 4))
    sns.barplot(x=list(metrics.keys()), y=list(metrics.values()), palette="Set2")
    plt.ylim(0, 1)
    plt.title("Pitstop Classification Metrics")
    plt.grid(True)
    plt.show()

# 4. Curvas de Entrenamiento
def plot_training_history(history, metric="accuracy"):
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.plot(history.history[metric], label=f"Train {metric}")
    plt.plot(history.history[f"val_{metric}"], label=f"Val {metric}")
    plt.title(f"{metric.capitalize()} over Epochs")
    plt.xlabel("Epochs")
    plt.ylabel(metric.capitalize())
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(history.history["loss"], label="Train Loss")
    plt.plot(history.history["val_loss"], label="Val Loss")
    plt.title("Loss over Epochs")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# 5. Importancia de Variables
def plot_feature_importance(X, y, feature_names, top_n=15):
    rf = RandomForestRegressor(n_estimators=100, random_state=42)
    rf.fit(X, y)
    importances = rf.feature_importances_
    indices = np.argsort(importances)[::-1][:top_n]
    plt.figure(figsize=(8, 6))
    sns.barplot(x=importances[indices], y=np.array(feature_names)[indices], palette="viridis")
    plt.title("Top Feature Importances")
    plt.xlabel("Importance")
    plt.ylabel("Feature")
    plt.grid(True)
    plt.show()

# 6. Embeddings 2D
def plot_embeddings_2d(embedding_matrix, labels, title="Embeddings (2D projection)"):
    pca = PCA(n_components=2)
    reduced = pca.fit_transform(embedding_matrix)
    plt.figure(figsize=(8, 6))
    plt.scatter(reduced[:, 0], reduced[:, 1], c=labels, cmap="tab20", alpha=0.7)
    plt.title(title)
    plt.xlabel("PCA 1")
    plt.ylabel("PCA 2")
    plt.grid(True)
    plt.colorbar()
    plt.show()
```
